```python
import pandas as pd
import os
```

```python
orders = pd.read_csv("data/raw/olist_orders_dataset.csv")

date_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_cols:
    orders[col] = pd.to_datetime(orders[col], errors='coerce')

orders = orders[
    orders['order_status'].isin(
        ['delivered', 'shipped', 'processing', 'approved']
    )
]

orders = orders.dropna(subset=['order_id'])

os.makedirs("data/processed", exist_ok=True)

orders.to_csv(
    "data/processed/orders_cleaned.csv",
    index=False
)

print("Orders cleaned and saved")
```

```python
items = pd.read_csv("data/raw/olist_order_items_dataset.csv")

items = items[items['price'] > 0]
items = items[items['freight_value'] >= 0]

items['total_value'] = items['price'] + items['freight_value']

items.to_csv(
    "data/processed/order_items_cleaned.csv",
    index=False
)

print("Order items cleaned and saved")
```

```python
products = pd.read_csv("data/raw/olist_products_dataset.csv")

products['product_category_name'] = products[
    'product_category_name'
].fillna('unknown')

products.to_csv(
    "data/processed/products_cleaned.csv",
    index=False
)

print("Products cleaned and saved")
```
